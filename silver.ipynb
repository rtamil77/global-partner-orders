```python
agg_df = df_order_items.groupBy('APP_NAME').agg(F.count('APP_NAME').alias('cnt'))
agg_df.display()
```

```python
df_order_items = df_order_items.filter(~F.col('APP_NAME').isin('JUICE"'))
print(df_order_items.count())
#df_order_items.display() ,'Alltown Fresh - DEVELOPMENT'
```

```python
agg_df = df_order_items.groupBy('RESTAURANT_ID').agg(F.count('RESTAURANT_ID').alias('cnt'))
agg_df = agg_df.withColumn('restaurant_len',F.length(F.col('RESTAURANT_ID')))
agg_df.display()
sum = agg_df.agg(F.sum('cnt')).first()
print(sum)
```

```python
agg_df = df_order_items.groupBy('ORDER_ID').agg(F.count('*').alias('cnt')).orderBy(F.col('cnt'),ascending=False)
agg_df.display()
sum_order_id = agg_df.agg(F.sum(F.col('cnt'))).first()
print(sum_order_id)
cnt_order_id = agg_df.agg(F.count(F.col('cnt'))).first()
print(cnt_order_id)
```
